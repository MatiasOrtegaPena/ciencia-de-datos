```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 100)
amplitude = np.sin(t)

plt.plot(t, amplitude)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude (arb. units)")
plt.title("Sine wave over 10 s")
plt.show()
```
